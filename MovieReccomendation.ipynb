# Reccomendation System

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizor=TfidfVectorizer()
```

```python
Model=vectorizor.fit_transform(data[['Title','Actors 1', 'Actors 2','Actors 3', 'Director','Genre', 'Music','Producer']])
```

```python
type(Model)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(Model)
```

```python

selected_cols = ['Title', 'Actors 1', 'Actors 2', 'Actors 3', 'Director', 'Genre', 'Music', 'Producer']


data[selected_cols] = data[selected_cols].fillna('')


data['combined'] = data[selected_cols].astype(str).agg(' '.join, axis=1)


vectorizer = TfidfVectorizer()
Model = vectorizer.fit_transform(data['combined'])


similarity = cosine_similarity(Model)

def recommend(movie_title):
    movie_title = movie_title.lower()
    
    # Title case sensitive 
    index = data[data['Actors 1'].str.lower() == movie_title].index.tolist()
    
    if len(index) == 0:
        return "Movie not found in dataset."

    index = index[0]

    # Similarity scores
    scores = list(enumerate(similarity[index]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Top 5 similar movies
    top_movies = scores[1:6]

    print(f"\nTop 5 recommendations for '{data.iloc[index]['Title']}':\n")
    for i, score in top_movies:
        print(f"{data.iloc[i]['Title']} (Score: {round(score, 2)})")

recommend("Akshay Kumar")  
```
